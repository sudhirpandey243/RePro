```python
# ════════════════════════════════════════════════════════════════════════════════════════════
#  MSCOCO CELL — SA-AET vs SA-AET + proj_univ on an MSCOCO test subset (SA-AET Table IV protocol)
#  One Colab cell, self-contained (same environment / fixes as the Flickr v4 cell).
#
#  EDIT ONLY:  SOURCE = "CLIP_ViT" | "ALBEF" | "TCL" | "CLIP_CNN"      (one per Colab session)
#
#  * test images: the FIRST N_IMAGES (= 1,000) entries of SA-AET's MSCOCO 5K test list (coco_test.json);
#    copied from your Drive folder COCO_IMG; any that are not there are downloaded ONCE from images.cocodataset.org
#    into COCO_EXTRA on Drive. Clean-rank files are computed for this subset (smaller gallery than the paper's 5K,
#    so numbers are NOT directly comparable with Table IV; SA-AET vs ours in the same run IS).
#  * models: SA-AET's MSCOCO checkpoints (albef_mscoco.pth, tcl_mscoco.pth) + OpenAI CLIP.
#  * your method: blocks 7-10, lambda 4, rank 1 = the frozen Flickr setting, NOT re-tuned on COCO. The register
#    basis is rebuilt for the COCO checkpoints from 96 NON-test COCO images (every image in SA-AET's COCO test /
#    val lists excluded) -> fixes the possible test overlap of the old MSCOCO pilot basis.
#  * runs: SA-AET, SA-AET + proj_univ (lambda 4), optional flat lambda 4; per-image paired test; images saved.
#  Time on an A100 (1,000 images): CLIP_ViT ~1.5 h per run, ALBEF / TCL ~3 h per run (COCO re-ranks top-256).
#  Finished runs are saved on Drive and reused; a disconnect loses at most the run in progress.
# ════════════════════════════════════════════════════════════════════════════════════════════
import datetime, glob, json, os, re, shutil, subprocess, sys, zipfile
from pathlib import Path

# ── 0. self-contained start: this cell sets up everything it needs itself ────────────────────
#    Only requirement: CELL 1 of sa_aet_phase1_colab.py has run at least once in THIS Colab runtime
#    (it installs the Python 3.10 environment and the SA-AET code into /content, which a restart erases).
REPO = Path("/content/SA-AET")
ENV_PY = "/content/sa_env/bin/python"
OUT = Path("/content/drive/MyDrive/sa_aet_phase1")
try:
    from google.colab import drive
    if not os.path.isdir("/content/drive/MyDrive"):
        drive.mount("/content/drive")
except ImportError:
    pass
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"


def sh(cmd, log=None):
    """run a shell command, stream its output, optionally tee it to a log file; stop on failure."""
    print(f"$ {cmd}")
    p = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    f = open(log, "a") if log else None
    for line in p.stdout:
        print(line, end="")
        if f:
            f.write(line); f.flush()
    p.wait()
    if f:
        f.close()
    if p.returncode != 0:
        raise SystemExit(f"command failed (exit {p.returncode}): {cmd}")


# ── 0. setup (automatic; skipped when already done in this runtime, ~10 min the first time) ──
import hashlib, urllib.request
sh("nvidia-smi --query-gpu=name,memory.total --format=csv")
# 0a. SA-AET code
if not REPO.exists():
    sh(f"git clone -q https://github.com/jiaxiaojunQAQ/SA-AET.git {REPO}")

# 0b. FIX 1 — import bug: eval_AET.py / eval.py import paired_dataset but use paired_dataset2
for f in ("eval_AET.py", "eval.py"):
    p = REPO / f
    s = p.read_text()
    s2 = re.sub(r"^from dataset import paired_dataset\s*$", "from dataset import paired_dataset, paired_dataset2",
                s, flags=re.M)
    if s2 != s:
        p.write_text(s2); print(f"  patched import in {f}")

# 0c. FIX 2 — out-of-memory: SA-AET encodes ~10,000 corpus captions in ONE forward pass, for EVERY batch.
#    Encode them once, 256 at a time, and reuse them. Same features, far less memory and time.
_p = REPO / "SA_AET.py"; _s = _p.read_text()
_OLD = """            all_texts_input = self.txt_attacker.tokenizer(all_texts, padding='max_length', truncation=True,
                                                     max_length=max_length, return_tensors="pt").to(device)
            all_texts_output = self.model.inference_text(all_texts_input)
            all_txt_supervisions = all_texts_output['text_feat']
"""
_NEW = """            if getattr(self, "_all_txt_key", None) != (id(all_texts), len(all_texts)):   # PATCHED: encode once, chunked
                _feats = []
                for _i in range(0, len(all_texts), 256):
                    _inp = self.txt_attacker.tokenizer(all_texts[_i:_i + 256], padding='max_length', truncation=True,
                                                       max_length=max_length, return_tensors="pt").to(device)
                    _feats.append(self.model.inference_text(_inp)['text_feat'])
                self._all_txt_cache = torch.cat(_feats)
                self._all_txt_key = (id(all_texts), len(all_texts))
            all_txt_supervisions = self._all_txt_cache
"""
if "PATCHED: encode once" not in _s:
    if _OLD not in _s:
        raise SystemExit("could not find the text-encoding block in SA_AET.py")
    _p.write_text(_s.replace(_OLD, _NEW)); print("  patched SA_AET.py (caption corpus encoded once)")

# 0d. separate Python 3.10 environment with the versions the repo was written for
if not Path(ENV_PY).exists():
    sh(f"{sys.executable} -m pip install -q uv")
    sh("uv venv /content/sa_env --python 3.10 -q")
    sh(f"uv pip install -q --python {ENV_PY} torch==2.1.0 torchvision==0.16.0 "
       f"--index-url https://download.pytorch.org/whl/cu121")
    sh(f"uv pip install -q --python {ENV_PY} transformers==4.28.1 timm==0.6.13 ftfy==6.1.1 "
       f"ruamel.yaml==0.17.21 'numpy<2' regex tqdm pillow setuptools==69.5.1 huggingface_hub==0.16.4")
sh(f"{ENV_PY} -c \"import torch, transformers, timm; print('torch', torch.__version__, 'cuda', "
   f"torch.cuda.is_available(), '| transformers', transformers.__version__, '| timm', timm.__version__)\"")

# 0e. OpenAI CLIP weights (ViT-B/16, RN101) into ~/.cache/clip; try two hosts.
clip_cache = Path.home() / ".cache/clip"; clip_cache.mkdir(parents=True, exist_ok=True)
CLIP_FILES = {"ViT-B-16.pt": "5806e77cd80f8b59890b7e101eabd078d9fb84e6937f9e85e4ecb61988df416f",
              "RN101.pt":    "8fa8567bab74a42d41c5915025a8e4538c3bdbe8804a470a72f30b0d94fab599"}
for fn, sha in CLIP_FILES.items():
    dst = clip_cache / fn
    ok = dst.exists() and hashlib.sha256(dst.read_bytes()).hexdigest() == sha
    for host in ("https://openaipublic.azureedge.net", "https://openaipublic.blob.core.windows.net"):
        if ok:
            break
        try:
            urllib.request.urlretrieve(f"{host}/clip/models/{sha}/{fn}", dst)
            ok = hashlib.sha256(dst.read_bytes()).hexdigest() == sha
        except Exception as e:                                   # noqa: BLE001
            print(f"  {host}: {e}")
    print(f"  CLIP {fn}: {'OK' if ok else 'FAILED — download it manually into ' + str(clip_cache)}")


SOURCE = "CLIP_CNN"      # order (cheapest first): "CLIP_ViT" -> "ALBEF" -> "TCL" -> "CLIP_CNN"
N_IMAGES = 1000          # first N of SA-AET's 5K MSCOCO test list (None = all 5,000: ~5x the time)
RUN_FLAT = False         # True: also run flat lambda 4 (ablation on COCO, +1 run)
SAVE_IMAGES = True       # adversarial PNGs + captions -> Drive (one zip per run)
SKIP_SOURCE_ITM = True   # white-box column ITC-only (saves time; the attack is unchanged)
LAM, LAYERS = 4.0, "7,8,9,10"
RUNS = [("SA-AET", "none", "", 0.0, 42), (f"proj_L7-10_lam{LAM:g}", "proj", LAYERS, LAM, 42)] \
       + ([(f"flat_L7-10_lam{LAM:g}", "flat", LAYERS, LAM, 42)] if RUN_FLAT else [])
COCO_IMG = "/content/drive/MyDrive/datasets/coco2014/val2014"            # your COCO val2014 folder
COCO_ANN = "/content/drive/MyDrive/datasets/captions_val2014.json"
COCO_EXTRA = Path("/content/drive/MyDrive/coco2014/coco_test_extra")   # test images not in COCO_IMG go here
LOCAL_ROOT = Path("/content/coco_local")                               # fast local copy: <root>/val2014/<file>
FOUT = OUT / "coco"; FOUT.mkdir(parents=True, exist_ok=True)
BASIS_COCO = FOUT / f"register_basis_coco_{SOURCE}.pt"
if not os.path.isdir(COCO_IMG):
    raise SystemExit(f"COCO folder not found: {COCO_IMG} — fix COCO_IMG")
LI = None

# 0f. the attack module (SA-AET ImageAttacker + IFD, with or without projection)
REG_ATTACK = r'''
""" SA-AET image attack + intermediate-feature disruption (IFD), with or without projection.
    L = L_SA-AET - lam * sum_l w_l cos( P_l(F_l(x_adv)), P_l(F_l(x_clean)) ),   sum_l w_l = 1
        flat: P_l = I      proj: P_l = I - U_l U_l^T (register direction)      rand: same with a random U_l
        w_l: equal by default (= mean over layers, the original setting) or from --ifd_weights
        a layer without a register direction in the basis uses P_l = I (plain cosine), as in the main code
        U_l may have 1 or 2 columns (rank-1 / rank-2 basis); the basis file decides
        alpha (optional, idea 3 "projection + norm tempering"): instead of the flattened cosine, use
            sum_i w_i cos(P a_i, P c_i),  w_i ∝ ||P c_i||^alpha  (clean projected token norms, no gradient)
        alpha = None -> flattened cosine (the original proj_univ, ~ token weights ||a_i|| ||c_i||)
    The IFD term is added inside loss_func, so it also drives SA-AET's triangle-sample selection. """
import torch
import torch.nn.functional as F
from torchvision import transforms
from SA_AET import ImageAttacker

SA_SCALES = [0.5, 0.75, 1.25, 1.5]              # hard-coded inside SA-AET's attack
RAND_SEED = 1234                                # fixed: the same random directions in every run


def vision_blocks(model):
    if hasattr(model, "visual") and hasattr(model.visual, "transformer"):
        return model.visual.transformer.resblocks, True          # OpenAI CLIP ViT: [L, N, D]
    if hasattr(model, "visual_encoder"):
        return model.visual_encoder.blocks, False                 # ALBEF / TCL: [N, L, D]
    raise SystemExit("this source model has no ViT blocks (CLIP_CNN): use --ifd_mode none")


class RegImageAttacker(ImageAttacker):
    def __init__(self, model, normalization, mode, layers, lam, basis=None, weights=None, alpha=None, **kw):
        super().__init__(normalization, **kw)
        if mode not in ("flat", "proj", "rand"):
            raise SystemExit(f"--ifd_mode must be none, flat, proj or rand (got {mode})")
        self.mode, self.layers, self.lam = mode, list(layers), float(lam)
        self.alpha = None if alpha is None else float(alpha)
        w = [1.0] * len(self.layers) if not weights else [float(v) for v in weights]
        if len(w) != len(self.layers) or sum(w) <= 0:
            raise SystemExit(f"--ifd_weights must have one positive-sum value per layer (got {weights})")
        self.w = torch.tensor(w) / sum(w)                        # sum to 1 -> lam keeps the same scale
        blocks, self.lnd = vision_blocks(model)
        self.feats, self.slice_i, self.b = {}, 0, None
        self.handles = [blocks[l].register_forward_hook(self._hook(l)) for l in self.layers]
        self.U = {}
        for l in self.layers:
            if mode == "flat":
                continue
            if basis is None or basis.get(l) is None:            # no registers in this block -> plain cosine
                print(f"  block {l + 1}: no register direction in the basis -> plain cosine there")
                continue
            U = basis[l].float()
            if mode == "rand":                                   # random unit direction, same shape as U
                g = torch.Generator().manual_seed(RAND_SEED + l)
                U = F.normalize(torch.randn(U.shape, generator=g), dim=0)
            self.U[l] = U
        print(f"  RegImageAttacker: mode={mode}, layers={[l + 1 for l in self.layers]}, "
              f"weights={[round(float(v), 3) for v in self.w]}, lam={self.lam}, "
              f"rank={[int(self.U[l].shape[1]) if l in self.U else 0 for l in self.layers]}, alpha={self.alpha}")

    def close(self):                                             # remove the hooks (needed when re-using the model)
        for h in self.handles:
            h.remove()
        self.handles = []

    def _hook(self, l):
        def fn(module, inputs, output):
            t = output[0] if isinstance(output, tuple) else output
            self.feats[l] = t.permute(1, 0, 2) if self.lnd else t
            self.slice_i = 0
        return fn

    def _proj(self, x, l):
        if self.mode == "flat" or l not in self.U:
            return x
        U = self.U[l].to(device=x.device, dtype=x.dtype)
        return x - (x @ U) @ U.T

    def _ifd(self, rows, ref):
        cs = []
        for l in self.layers:
            a = self._proj(self.feats[l][rows].float(), l)
            c = self._proj(ref[l].float(), l)
            if self.alpha is None:                               # original: flattened cosine
                cs.append(F.cosine_similarity(a.flatten(1), c.flatten(1), dim=-1))
            else:                                                # norm-tempered per-token cosine
                wt = c.detach().norm(dim=-1).clamp(min=1e-6).pow(self.alpha)
                wt = wt / wt.sum(1, keepdim=True)
                cs.append((wt * F.cosine_similarity(a, c, dim=-1)).sum(1))
        w = self.w.to(cs[0].device)                              # equal weights == the original mean
        return (torch.stack(cs) * w[:, None]).sum(0).mean()

    def _clean_views(self, imgs):
        H, W = imgs.shape[-2:]
        back = transforms.Resize((H, W), interpolation=transforms.InterpolationMode.BICUBIC)
        out = [imgs]
        for r in SA_SCALES:
            down = transforms.Resize((int(r * H), int(r * W)), interpolation=transforms.InterpolationMode.BICUBIC)
            out.append(back(torch.clamp(down(imgs), 0.0, 1.0)))
        return torch.cat(out, 0)

    def txt_guided_attack(self, model, imgs, txt2img, all_txt_supervisions, device, scales=None, txt_embeds=None):
        self.b = imgs.shape[0]
        with torch.no_grad():
            model.inference_image(self.normalization(self._clean_views(imgs)))
            self.ref = {l: self.feats[l].detach().float().clone() for l in self.layers}
        return super().txt_guided_attack(model, imgs, txt2img, all_txt_supervisions, device,
                                         scales=scales, txt_embeds=txt_embeds)

    def loss_func(self, adv_imgs_embeds, txts_embeds, txt2img, all_txt_supervisions):
        loss = super().loss_func(adv_imgs_embeds, txts_embeds, txt2img, all_txt_supervisions)
        if self.b is None or not self.feats:
            return loss
        n, b = next(iter(self.feats.values())).shape[0], self.b
        if n == b:
            ifd = self._ifd(slice(0, b), {l: self.ref[l][:b] for l in self.layers})
        elif n == 5 * b:
            i = self.slice_i % 5
            self.slice_i += 1
            ifd = self._ifd(slice(i * b, (i + 1) * b), {l: self.ref[l][i * b:(i + 1) * b] for l in self.layers})
        else:
            return loss
        return loss - self.lam * ifd
'''
(REPO / "reg_attack.py").write_text(REG_ATTACK)

# 2. eval_reg.py = eval_AET.py + options --ifd_mode, --ifd_layers, --lam, --basis (always rebuilt from eval_AET.py)
src = (REPO / "eval_AET.py").read_text()
src = src.replace("from SA_AET import Attacker, ImageAttacker, TextAttacker",
                  "from SA_AET import Attacker, ImageAttacker, TextAttacker\nfrom reg_attack import RegImageAttacker", 1)
src = src.replace(
    "    img_attacker = ImageAttacker(images_normalize, eps=8/255, steps=10, step_size=2/255)",
    "    if args.ifd_mode == 'none':\n"
    "        img_attacker = ImageAttacker(images_normalize, eps=8/255, steps=10, step_size=2/255)\n"
    "    else:\n"
    "        _basis = (torch.load(args.basis, map_location='cpu')['basis'][args.source_model]\n"
    "                  if args.ifd_mode in ('proj', 'rand') else None)\n"
    "        _w = [float(v) for v in args.ifd_weights.split(',')] if args.ifd_weights else None\n"
    "        img_attacker = RegImageAttacker(model, images_normalize, args.ifd_mode,\n"
    "                                        [int(v) - 1 for v in args.ifd_layers.split(',')], args.lam, _basis,\n"
    "                                        weights=_w, alpha=(None if args.ifd_alpha == '' else float(args.ifd_alpha)),\n"
    "                                        eps=8/255, steps=10, step_size=2/255)", 1)
src = src.replace(
    "    parser.add_argument('--scales', type=str, default='0.5,0.75,1.25,1.5')",
    "    parser.add_argument('--scales', type=str, default='0.5,0.75,1.25,1.5')\n"
    "    parser.add_argument('--ifd_mode', type=str, default='none')         # none | flat | proj | rand\n"
    "    parser.add_argument('--ifd_layers', type=str, default='7,8,9,10')   # 1-based block numbers\n"
    "    parser.add_argument('--ifd_weights', type=str, default='')          # one weight per layer; '' = equal\n"
    "    parser.add_argument('--ifd_alpha', type=str, default='')            # '' = flattened cosine (original)\n"
    "    parser.add_argument('--lam', type=float, default=1.0)\n"
    "    parser.add_argument('--basis', type=str, default='')", 1)
assert src.count("RegImageAttacker") == 2 and "--ifd_mode" in src, "patching eval_AET.py failed — send me eval_AET.py"
# v4: save per-image results (which queries are still correct after the attack) for the paired test
_IE = "    eval_result = {'txt_r1_ASR (txt_r1)': f'{asr_tr1}({tr1})',"
assert src.count(_IE) == 1, "itm_eval patch failed — send me eval_AET.py"
src = src.replace(_IE,
    "    if os.environ.get('PERIMG_DIR'):\n"
    "        os.makedirs(os.environ['PERIMG_DIR'], exist_ok=True)\n"
    "        np.savez(os.path.join(os.environ['PERIMG_DIR'], f'{model_name}.npz'), tr1_ok=after_attack_tr1,\n"
    "                 ir1_ok=after_attack_ir1, n_img=scores_i2t.shape[0], n_txt=scores_t2i.shape[0],\n"
    "                 txt2img=np.array([txt2img[i] for i in range(scores_t2i.shape[0])]))\n" + _IE, 1)
# v4: optional — skip the slow ITM re-ranking for the SOURCE model only (white-box column; attack unchanged)
_SRC = ("    if args.source_model in ['ALBEF', 'TCL']:\n"
        "        s_score_matrix_i2t, s_score_matrix_t2i = retrieval_score(model,")
assert src.count(_SRC) == 1, "source-ITM patch failed — send me eval_AET.py"
src = src.replace(_SRC, _SRC.replace("['ALBEF', 'TCL']:", "['ALBEF', 'TCL'] and not os.environ.get('SKIP_SOURCE_ITM'):"), 1)
# v4: save adversarial images (PNG, attack resolution) + adversarial captions
_AT = ("        adv_images, adv_texts,execuate_time = attacker.attack(images, texts, txt2img,all_texts, device=device,\n"
       "                                                max_length=max_length, scales=scales)\n")
assert src.count(_AT) == 1, "attack-call patch failed — send me eval_AET.py"
src = src.replace(_AT, _AT + "        if os.environ.get('ADV_DIR'):\n"
                              "            _save_adv(adv_images, image_paths, adv_texts, os.environ['ADV_DIR'])\n", 1)
src = src.replace("from reg_attack import RegImageAttacker",
    "from reg_attack import RegImageAttacker\n\n"
    "def _save_adv(adv_images, image_paths, adv_texts, out_dir):\n"
    "    os.makedirs(out_dir, exist_ok=True)\n"
    "    for k, p in enumerate(image_paths):\n"
    "        arr = (adv_images[k].detach().clamp(0, 1).permute(1, 2, 0).cpu().numpy() * 255).round().astype('uint8')\n"
    "        Image.fromarray(arr).save(os.path.join(out_dir, os.path.splitext(os.path.basename(p))[0] + '.png'))\n"
    "    with open(os.path.join(out_dir, 'adv_captions.jsonl'), 'a') as f:\n"
    "        f.write(json.dumps({'images': [os.path.basename(p) for p in image_paths], 'adv_texts': list(adv_texts)}) + '\\n')\n", 1)
(REPO / "eval_reg.py").write_text(src)
print("  wrote reg_attack.py and eval_reg.py")

# ── A1. MSCOCO test images: the first N_IMAGES of SA-AET's 5K test list ─────────────────────
ann_all = json.load(open(REPO / "data_annotation/coco_test.json"))
use_ann = ann_all[:N_IMAGES] if N_IMAGES else ann_all
N_BATCHES = len(use_ann) if N_IMAGES else None
SUB = f"sub{len(use_ann)}" if N_IMAGES else "full"
need = [a["image"].split("/")[-1] for a in use_ann]                       # COCO_val2014_000000xxxxxx.jpg
COCO_EXTRA.mkdir(parents=True, exist_ok=True)
on_drive = set(os.listdir(COCO_IMG))
missing = [n for n in need if n not in on_drive and not (COCO_EXTRA / n).exists()]
print(f"  MSCOCO test images: {len(need) - len(missing)} / {len(need)} already on Drive")
if missing:
    print(f"  downloading {len(missing)} images from images.cocodataset.org (once; kept in {COCO_EXTRA}) ...")
    import concurrent.futures as _cf, urllib.request as _ur

    def _get(n):
        for url in (f"http://images.cocodataset.org/val2014/{n}", f"https://images.cocodataset.org/val2014/{n}"):
            try:
                _ur.urlretrieve(url, COCO_EXTRA / n)
                if (COCO_EXTRA / n).stat().st_size > 1000:
                    return True
            except Exception:                                    # noqa: BLE001
                pass
        return False
    with _cf.ThreadPoolExecutor(16) as _ex:
        _ok = list(_ex.map(_get, missing))
    print(f"  downloaded {sum(_ok)} / {len(missing)}")
    if not all(_ok):
        raise SystemExit(f"{len(_ok) - sum(_ok)} test images could not be downloaded — send me this output")
(LOCAL_ROOT / "val2014").mkdir(parents=True, exist_ok=True)               # coco_test.json paths are val2014/<file>
for n in need:
    dst = LOCAL_ROOT / "val2014" / n
    if not dst.exists():
        shutil.copy((Path(COCO_IMG) / n) if n in on_drive else (COCO_EXTRA / n), dst)
print(f"  local copy ready: {LOCAL_ROOT} ({len(need)} images)")

# ── A2. MSCOCO checkpoints + config ─────────────────────────────────────────────────────────
ck = REPO / "checkpoints"; ck.mkdir(exist_ok=True)
for name in ("albef_mscoco.pth", "tcl_mscoco.pth"):
    if not (ck / name).exists() or (ck / name).stat().st_size < 10_000_000:
        sh(f"wget -q -O {ck / name} https://huggingface.co/Sensen02/VLPTransferAttackCheckpoints/resolve/main/{name}")
    if (ck / name).stat().st_size < 10_000_000:
        raise SystemExit(f"{name}: download failed or file too small — send me this output")
    print(f"  {name}: {(ck / name).stat().st_size / 1e6:.0f} MB")
y = (REPO / "configs/Retrieval_coco.yaml").read_text()
y = re.sub(r"^image_root:.*$", f"image_root: '{LOCAL_ROOT}/'", y, flags=re.M)
if N_IMAGES:
    json.dump(use_ann, open(REPO / f"data_annotation/coco_{SUB}.json", "w"))
    y = re.sub(r"^test_file:.*$", f"test_file: './data_annotation/coco_{SUB}.json'", y, flags=re.M)
    y = re.sub(r"^k_test:.*$", f"k_test: {min(256, len(use_ann))}", y, flags=re.M)   # ALBEF/TCL re-rank top-k
    RUN_CFG_F, IDX_DIR_F = f"./configs/coco_{SUB}.yaml", f"./std_eval_idx/mscoco_{SUB}"
else:
    y = re.sub(r"^test_file:.*$", "test_file: './data_annotation/coco_test.json'", y, flags=re.M)
    RUN_CFG_F, IDX_DIR_F = "./configs/coco_full.yaml", "./std_eval_idx/mscoco/"
(REPO / RUN_CFG_F).write_text(y)

# clean-rank files for the subset: which queries each model gets right on CLEAN images (needed for ASR)
CLEAN_IDX_SCRIPT = r'''
import os, sys, numpy as np, torch
from ruamel.yaml import YAML
from torchvision import transforms
from PIL import Image
import eval_AET as E
from dataset import paired_dataset2
from models import clip

cfg_path, albef_ckpt, tcl_ckpt, out_dir = sys.argv[1:5]
config = YAML(typ="safe").load(open(cfg_path))
E.config = config
args = type("A", (), {"albef_ckpt": albef_ckpt, "tcl_ckpt": tcl_ckpt})()
device = torch.device("cuda")
norm = transforms.Normalize((0.48145466, 0.4578275, 0.40821073), (0.26862954, 0.26130258, 0.27577711))
os.makedirs(out_dir, exist_ok=True)
for name in ["ALBEF", "TCL", "CLIP_ViT", "CLIP_CNN"]:
    model, _, tok = E.load_model(args, name, "bert-base-uncased", device)
    model = model.to(device).float().eval()
    if name in ["ALBEF", "TCL"]:
        tf = transforms.Compose([transforms.Resize((config["image_res"], config["image_res"]),
                                                   interpolation=Image.BICUBIC), transforms.ToTensor()])
    else:
        n = model.visual.input_resolution
        tf = transforms.Compose([transforms.Resize(n, interpolation=Image.BICUBIC), transforms.CenterCrop(n),
                                 transforms.ToTensor()])
    ds = paired_dataset2(config["test_file"], tf, config["image_root"])
    n_img, n_txt = len(ds), len(ds.text)
    with torch.no_grad():
        ifeat, iemb = [], []
        for i in range(0, n_img, 16):
            x = norm(torch.stack([ds[j][0] for j in range(i, min(i + 16, n_img))]).to(device))
            o = model.inference_image(x)
            ifeat.append(o["image_feat"].float().cpu())
            if name in ["ALBEF", "TCL"]:
                iemb.append(o["image_embed"].float().cpu())
        ifeat = torch.cat(ifeat)
        tfeat, temb, tatt = [], [], []
        for i in range(0, n_txt, 256):
            chunk = ds.text[i:i + 256]
            if name in ["ALBEF", "TCL"]:
                t = tok(chunk, padding="max_length", truncation=True, max_length=30, return_tensors="pt").to(device)
                o = model.inference_text(t)
                tfeat.append(o["text_feat"].float().cpu()); temb.append(o["text_embed"].float().cpu())
                tatt.append(t.attention_mask.cpu())
            else:
                e = model.encode_text(clip.tokenize(chunk, 77, True).to(device))
                tfeat.append(torch.nn.functional.normalize(e, dim=-1).float().cpu())
        tfeat = torch.cat(tfeat)
        if name in ["ALBEF", "TCL"]:
            s_i2t, s_t2i = E.retrieval_score(model, ifeat, torch.cat(iemb), tfeat, torch.cat(temb), torch.cat(tatt),
                                             n_img, n_txt, device=device)
            s_i2t, s_t2i = s_i2t.cpu().numpy(), s_t2i.cpu().numpy()
        else:
            sim = (ifeat @ tfeat.t()).numpy()
            s_i2t, s_t2i = sim, sim.T
    tr = np.zeros(n_img)
    for i, sc in enumerate(s_i2t):
        order = np.argsort(sc)[::-1]
        tr[i] = min(np.where(order == t)[0][0] for t in ds.img2txt[i])
    ir = np.array([np.where(np.argsort(sc)[::-1] == ds.txt2img[i])[0][0] for i, sc in enumerate(s_t2i)])
    for k in (1, 5, 10):
        np.save(f"{out_dir}/{name}_tr{k}_rank_index.npy", np.where(tr < k)[0])
        np.save(f"{out_dir}/{name}_ir{k}_rank_index.npy", np.where(ir < k)[0])
    print(f"  {name}: clean TR R@1 {100 * (tr < 1).mean():.1f}  IR R@1 {100 * (ir < 1).mean():.1f}")
    del model; torch.cuda.empty_cache()
print("clean rank files written to", out_dir)
'''
if not (REPO / IDX_DIR_F / "CLIP_CNN_ir10_rank_index.npy").exists():
    (REPO / "make_clean_idx.py").write_text(CLEAN_IDX_SCRIPT)
    sh(f"cd {REPO} && {ENV_PY} make_clean_idx.py {RUN_CFG_F} ./checkpoints/albef_mscoco.pth "
       f"./checkpoints/tcl_mscoco.pth {IDX_DIR_F}", log=str(FOUT / f"clean_idx_{SUB}.txt"))
print(f"  image set: MSCOCO {SUB} ({len(use_ann)} images), config {RUN_CFG_F}, clean ranks {IDX_DIR_F}")

# ── A3. register basis for the COCO checkpoints (forward only; 96 NON-test COCO images; once per source) ──
BASIS_SCRIPT = r'''
""" Same procedure as the tuning cell: COCO val2014 minus SA-AET's COCO test/val lists, sorted, shuffled (seed 42),
    first 96 images; register = patch token with norm > 5 x median; rank-1 direction per block (>= 8 images). """
import json, os, random, sys
import torch, torch.nn.functional as F
from ruamel.yaml import YAML
from torchvision import transforms
from PIL import Image
import eval_AET as E

src, cfg_path, albef_ckpt, tcl_ckpt, img_dir, ann_file, out_file = sys.argv[1:8]
ref_file = sys.argv[8] if len(sys.argv) > 8 else ""
config = YAML(typ="safe").load(open(cfg_path)); E.config = config
args = type("A", (), {"albef_ckpt": albef_ckpt, "tcl_ckpt": tcl_ckpt})()
dev = torch.device("cuda")
norm = transforms.Normalize((0.48145466, 0.4578275, 0.40821073), (0.26862954, 0.26130258, 0.27577711))
excl = set()
for f in ("data_annotation/coco_test.json", "data_annotation/coco_karpathy_test.json",
          "data_annotation/coco_karpathy_val.json"):
    if os.path.exists(f):
        excl |= {os.path.basename(a["image"]) for a in json.load(open(f)) if isinstance(a, dict) and "image" in a}
anns = json.load(open(ann_file))
id2file = {im["id"]: im["file_name"] for im in anns["images"]}
has_cap = {id2file[a["image_id"]] for a in anns["annotations"]}
on_disk = set(os.listdir(img_dir))
cands = sorted(n for n in has_cap if n in on_disk and n not in excl)
random.Random(42).shuffle(cands)
imgs = cands[:96]
print(f"  {len(excl)} COCO test/val images excluded; basis from {len(imgs)} non-test images")
model, _, _ = E.load_model(args, src, "bert-base-uncased", dev)
model = model.to(dev).float().eval()
if hasattr(model, "visual_encoder"):
    blocks, lnd, res = model.visual_encoder.blocks, False, config["image_res"]
    tf = transforms.Compose([transforms.Resize((res, res), interpolation=Image.BICUBIC), transforms.ToTensor()])
else:
    blocks, lnd, res = model.visual.transformer.resblocks, True, model.visual.input_resolution
    tf = transforms.Compose([transforms.Resize(res, interpolation=Image.BICUBIC), transforms.CenterCrop(res),
                             transforms.ToTensor()])
feats = {}
def hook(l):
    def fn(m, i, o):
        t = o[0] if isinstance(o, tuple) else o
        feats[l] = t.permute(1, 0, 2) if lnd else t
    return fn
hs = [blocks[l].register_forward_hook(hook(l)) for l in range(len(blocks))]
rows, nimg = {l: [] for l in range(len(blocks))}, {l: 0 for l in range(len(blocks))}
with torch.no_grad():
    for i in range(0, len(imgs), 8):
        x = torch.stack([tf(Image.open(os.path.join(img_dir, n)).convert("RGB")) for n in imgs[i:i + 8]]).to(dev)
        model.inference_image(norm(x))
        for l in range(len(blocks)):
            p = feats[l][:, 1:].float(); n = p.norm(dim=-1)
            reg = n > 5.0 * n.median(dim=1, keepdim=True).values
            nimg[l] += int(reg.any(1).sum())
            if reg.any():
                rows[l].append(F.normalize(p[reg], dim=-1).cpu())
for h in hs:
    h.remove()
basis, cap = {}, {}
for l in range(len(blocks)):
    if nimg[l] >= 8:
        R = torch.cat(rows[l]); U = torch.linalg.svd(R, full_matrices=False).Vh[:1].T.contiguous()
        basis[l] = U; cap[l] = float(((R @ U) ** 2).sum(1).mean())
for l in sorted(basis):
    print(f"   block {l + 1:>2}: {nimg[l]:>3} images with registers, rank-1 capture {100 * cap[l]:.1f} %")
if ref_file and os.path.exists(ref_file):                     # sanity: same direction as the Flickr-tuned basis?
    ref = torch.load(ref_file, map_location="cpu")["basis"].get(src, {})
    for l in sorted(set(ref) & set(basis)):
        print(f"   block {l + 1:>2}: |cos(COCO basis, Flickr basis)| = "
              f"{abs(float(F.normalize(ref[l].float()[:, 0], dim=0) @ basis[l][:, 0])):.3f}")
torch.save({"basis": {src: basis}, "meta": {"source": src, "images": imgs, "tau": 5.0, "rank": 1,
            "n_with_registers": nimg, "capture": cap, "checkpoint": "mscoco"}}, out_file)
print("  basis saved:", out_file)
'''
if SOURCE != "CLIP_CNN" and not BASIS_COCO.exists():
    (REPO / "build_basis_coco.py").write_text(BASIS_SCRIPT)
    _ref = OUT / "tuning" / f"register_basis_tuned_{SOURCE}.pt"
    sh(f"cd {REPO} && {ENV_PY} build_basis_coco.py {SOURCE} {RUN_CFG_F} ./checkpoints/albef_mscoco.pth "
       f"./checkpoints/tcl_mscoco.pth '{COCO_IMG}' '{COCO_ANN}' {BASIS_COCO} {_ref}",
       log=str(FOUT / f"basis_{SOURCE}.txt"))
if SOURCE != "CLIP_CNN":
    import torch as _t
    _b = _t.load(BASIS_COCO, map_location="cpu")["basis"][SOURCE]
    _need = [int(v) - 1 for v in LAYERS.split(",")]
    print(f"  COCO basis for {SOURCE}: blocks {[l + 1 for l in sorted(_b)]} "
          f"(method uses {[l + 1 for l in _need]}; missing -> plain cosine there: {[l + 1 for l in _need if l not in _b]})")

# ── B. runs (reused if already on Drive) ─────────────────────────────────────────────────────
def parse(block):
    out = {}
    for name, body in re.findall(r"Performance on (\w+): \n (\{.*?\})", block):
        d = eval(body)
        out[name] = (float(re.findall(r"[\d.]+", d["txt_r1_ASR (txt_r1)"])[0]),
                     float(re.findall(r"[\d.]+", d["img_r1_ASR (img_r1)"])[0]))
    return out


def load_saved(src):
    R = {}
    for f in glob.glob(str(FOUT / f"comparison_COCO-*_{src}_{SUB}_*.json")):
        d = json.load(open(f))
        lab = d["label"]
        if lab == "proj_tuned":                                    # old cells: name it by its lambda
            lab = f"proj_tuned_lam{float(d['lam']):g}"
        R[(lab, d["seed"])] = {k: tuple(v) for k, v in d["asr"].items()}
        if d.get("perimg_dir"):
            PERIMG[(lab, d["seed"])] = d["perimg_dir"]
    return R


PERIMG = {}
saved = load_saved(SOURCE)
for label, mode, layers, lam, seed in RUNS:
    if (label, seed) in saved:
        print(f"  {SOURCE} {label} seed {seed}: already on Drive -> reused"); continue
    if mode != "none" and SOURCE == "CLIP_CNN":
        print(f"  {SOURCE} {label}: skipped (CNN, no ViT blocks/registers; only SA-AET runs = control)"); continue
    tag = f"COCO-{label}_s{seed}_{SOURCE}_{SUB}_{datetime.datetime.now().strftime('%Y%m%d_%H%M')}"
    res_file = REPO / "result.txt"
    before = res_file.read_text() if res_file.exists() else ""
    weights, cli_mode = "", mode
    if mode == "proj":
        extra = f"--ifd_layers {layers} --lam {lam} --basis {BASIS_COCO}"
    elif mode == "flat":
        extra = f"--ifd_layers {layers} --lam {lam} --basis {BASIS_COCO}"
    else:
        extra = ""
    perimg_local, adv_local = Path(f"/content/perimg_{tag}"), Path(f"/content/adv_{tag}")
    for d_ in (perimg_local, adv_local):
        shutil.rmtree(d_, ignore_errors=True); d_.mkdir(parents=True)
    env = f"PERIMG_DIR={perimg_local} " + (f"ADV_DIR={adv_local} " if SAVE_IMAGES else "")
    env += "SKIP_SOURCE_ITM=1 " if SKIP_SOURCE_ITM else ""
    print(f"  -> running {SOURCE} {label} on {SUB} (est. "
          f"{(3.0 if SOURCE in ('ALBEF', 'TCL') else 1.5) * len(use_ann) / 1000:.1f} h)")
    sh(f"cd {REPO} && {env}{ENV_PY} eval_reg.py --config {RUN_CFG_F} --source_model {SOURCE} "
       f"--batch_size 8 --albef_ckpt ./checkpoints/albef_mscoco.pth --tcl_ckpt ./checkpoints/tcl_mscoco.pth "
       f"--original_rank_index_path {IDX_DIR_F} --cuda_id 0 --seed {seed} --ifd_mode {cli_mode} {extra}",
       log=str(FOUT / f"log_{tag}.txt"))
    new = res_file.read_text()[len(before):]
    (FOUT / f"result_{tag}.txt").write_text(new)
    perimg_drive = FOUT / "perimg" / tag                 # per-image results (small) -> Drive
    shutil.copytree(perimg_local, perimg_drive, dirs_exist_ok=True)
    if SAVE_IMAGES and any(adv_local.iterdir()):          # adversarial PNGs + captions -> one zip on Drive
        (FOUT / "adv_images").mkdir(exist_ok=True)
        shutil.make_archive(str(FOUT / "adv_images" / tag), "zip", adv_local)
        print(f"  adversarial images -> {FOUT / 'adv_images' / (tag + '.zip')}")
    json.dump({"label": label, "mode": mode, "layers": layers, "weights": weights, "lam": lam, "seed": seed,
               "source": SOURCE, "subset": SUB, "asr": parse(new), "perimg_dir": str(perimg_drive),
               "clean_rank_dir": str(REPO / IDX_DIR_F)},
              open(FOUT / f"comparison_{tag}.json", "w"), indent=2)
    saved = load_saved(SOURCE)

# ── C. table: paper (Table III) vs your SA-AET run vs your method ────────────────────────────
PAPER = {}                                   # subset gallery: no paper row (not comparable)
R = load_saved(SOURCE)
T = [t for t in ["ALBEF", "TCL", "CLIP_ViT", "CLIP_CNN"] if t != SOURCE]
mean = lambda r: sum(r[t][k] for t in T for k in (0, 1)) / (2 * len(T))
rows = ([("SA-AET (paper, 1K)" + (" *" if N_BATCHES else ""), PAPER[SOURCE])] if SOURCE in PAPER else []) \
       + [(f"{l} (seed {s})", R[(l, s)]) for (l, s) in sorted(R)]
print(f"\n  MSCOCO {SUB}, source {SOURCE} — ASR at R@1 (%), TR / IR")
print("  " + f"{'method':<30}" + "".join(f"{t:>16}" for t in T + [SOURCE]) + f"{'mean transfer':>15}")
for name, r in rows:
    if all(t in r for t in T):
        print("  " + f"{name:<30}" + "".join(f"{r[t][0]:>7.2f} /{r[t][1]:>6.2f}" for t in T + [SOURCE])
              + f"{mean(r):>15.2f}")
if N_BATCHES:
    print("\n  (MSCOCO subset: smaller gallery than the paper's 5K, so compare rows with each other, not with Table IV)")
elif ("SA-AET", 42) in R and SOURCE in PAPER:
    d = [abs(R[("SA-AET", 42)][t][k] - PAPER[SOURCE][t][k]) for t in T for k in (0, 1)]
    print(f"\n  reproduction: your SA-AET vs the paper, largest difference = {max(d):.2f} points "
          f"({'OK' if max(d) <= 3 else 'larger than expected — send me the log'})")
if ("SA-AET", 42) in R:
    for (l, s) in sorted(R):
        if l != "SA-AET" and s == 42:
            print(f"  {l} vs SA-AET (same run setup): {mean(R[(l, s)]) - mean(R[('SA-AET', 42)]):+.2f} points (mean transfer)")
elif SOURCE in PAPER and not N_BATCHES:
    for (l, s) in sorted(R):
        if s == 42:
            print(f"  {l} vs SA-AET (PAPER's numbers, same 1K test set): {mean(R[(l, s)]) - mean(PAPER[SOURCE]):+.2f}"
                  f" points (mean transfer; no own SA-AET run -> no paired p-value)")
if SKIP_SOURCE_ITM and SOURCE in ("ALBEF", "TCL"):
    print(f"  wb*: the {SOURCE} (white-box) column of runs made with SKIP_SOURCE_ITM is ITC-only (no ITM re-ranking)")
if False:
    print(f"  projW uses blocks {LI['layers']} with weights {[round(w, 2) for w in LI['weights']]} "
          f"(curve from non-test COCO images, fixed before this run)")

# ── C2. the key ablation: same lambda, with vs without projection ─────────────────────────
_flat = [l for (l, s_) in R if l.startswith("flat_") and s_ == 42]
_proj = [l for (l, s_) in R if l.startswith("proj_L7-10_lam") and s_ == 42]
for fl in _flat:
    lam_f = fl.rsplit("lam", 1)[-1]
    pj = f"proj_L7-10_lam{lam_f}"
    if pj in _proj and ("SA-AET", 42) in R:
        g_f = mean(R[(fl, 42)]) - mean(R[("SA-AET", 42)]); g_p = mean(R[(pj, 42)]) - mean(R[("SA-AET", 42)])
        print(f"\n  ABLATION (lambda {lam_f}): flat {g_f:+.2f}  vs  proj {g_p:+.2f}  points over SA-AET "
              f"-> projection adds {g_p - g_f:+.2f} points")
        print("     per target (TR / IR, proj minus flat): " + "   ".join(
            f"{t} {R[(pj, 42)][t][0] - R[(fl, 42)][t][0]:+.1f} / {R[(pj, 42)][t][1] - R[(fl, 42)][t][1]:+.1f}" for t in T))

# ── D. paired per-image significance test vs SA-AET (runs that saved per-image results) ───
import numpy as np


def _perimg(run_dir, model, clean_dir):
    """per image: (successes, counts) for TR (one query per image) and IR (its captions)."""
    z = np.load(Path(run_dir) / f"{model}.npz")
    o_tr = np.load(Path(clean_dir) / f"{model}_tr1_rank_index.npy")
    o_ir = np.load(Path(clean_dir) / f"{model}_ir1_rank_index.npy")
    n_img, t2i = int(z["n_img"]), z["txt2img"]
    tr_s, tr_n, ir_s, ir_n = (np.zeros(n_img) for _ in range(4))
    tr_n[o_tr] = 1
    tr_s[o_tr] = (~np.isin(o_tr, z["tr1_ok"])).astype(float)
    np.add.at(ir_n, t2i[o_ir], 1)
    np.add.at(ir_s, t2i[o_ir], (~np.isin(o_ir, z["ir1_ok"])).astype(float))
    return (tr_s, tr_n), (ir_s, ir_n)


def paired_test(base_dir, meth_dir, clean_dir, targets, n_perm=10000, n_boot=10000, seed=0):
    tests = []                                            # (name, per-image diff of successes, per-image count)
    for t in targets:
        (bt, nt), (bi, ni) = _perimg(base_dir, t, clean_dir)
        (mt, _), (mi, _) = _perimg(meth_dir, t, clean_dir)
        tests += [(f"{t} TR", mt - bt, nt), (f"{t} IR", mi - bi, ni)]
    S = np.stack([d for _, d, _ in tests]); Nn = np.stack([n for _, _, n in tests]); tot = Nn.sum(1)
    obs = 100 * S.sum(1) / tot
    rng = np.random.default_rng(seed)
    flips = rng.choice([-1.0, 1.0], size=(n_perm, S.shape[1]))
    perm = 100 * (flips @ S.T) / tot                      # [n_perm, n_tests]
    p = ((np.abs(perm) >= np.abs(obs) - 1e-9).sum(0) + 1) / (n_perm + 1)
    p_mean = ((np.abs(perm.mean(1)) >= abs(obs.mean()) - 1e-9).sum() + 1) / (n_perm + 1)
    idx = rng.integers(0, S.shape[1], size=(n_boot, S.shape[1]))
    boot = 100 * S[:, idx].sum(2) / np.maximum(Nn[:, idx].sum(2), 1)   # [n_tests, n_boot]
    ci = np.percentile(boot, [2.5, 97.5], axis=1)
    ci_mean = np.percentile(boot.mean(0), [2.5, 97.5])
    order = np.argsort(p); holm = np.empty_like(p); run_max = 0.0     # Holm step-down
    for r, i in enumerate(order):
        run_max = max(run_max, min(1.0, (len(p) - r) * p[i])); holm[i] = run_max
    return tests, obs, ci, p, holm, obs.mean(), ci_mean, p_mean


if ("SA-AET", 42) in PERIMG:
    for (l, s_) in sorted(PERIMG):
        if l == "SA-AET" or s_ != 42:
            continue
        try:
            tests, obs, ci, p, holm, gm, cim, pm = paired_test(PERIMG[("SA-AET", 42)], PERIMG[(l, s_)],
                                                               REPO / IDX_DIR_F, T)
        except FileNotFoundError as e:
            print(f"  paired test {l}: per-image file missing ({e})"); continue
        print(f"\n  PAIRED TEST  {l}  vs  SA-AET   ({SUB}; images resampled, captions stay with their image)")
        print(f"   {'target':<14}{'gain':>8}   {'95% CI':<18}{'p':>8}{'p (Holm)':>10}")
        for k, (nm, _, _) in enumerate(tests):
            print(f"   {nm:<14}{obs[k]:>+8.2f}   [{ci[0, k]:+6.2f}, {ci[1, k]:+6.2f}]  {p[k]:>8.4f}{holm[k]:>10.4f}")
        print(f"   {'MEAN transfer':<14}{gm:>+8.2f}   [{cim[0]:+6.2f}, {cim[1]:+6.2f}]  {pm:>8.4f}")
        json.dump({"source": SOURCE, "subset": SUB, "method": l, "tests": [t[0] for t in tests],
                   "gain": obs.tolist(), "ci_low": ci[0].tolist(), "ci_high": ci[1].tolist(), "p": p.tolist(),
                   "p_holm": holm.tolist(), "mean_gain": float(gm), "mean_ci": cim.tolist(), "mean_p": float(pm)},
                  open(FOUT / f"paired_{SOURCE}_{SUB}_{l}.json", "w"), indent=2)
else:
    print("\n  (paired test: needs SA-AET and a method run made with this v4 cell, which saves per-image results;"
          " older runs only have the overall numbers)")

# ── E. PAPER TABLE: every source finished so far on this image set (Table III layout) ──────
ALL_SRC = ["ALBEF", "TCL", "CLIP_ViT", "CLIP_CNN"]
print(f"\n  ══ PAPER TABLE — MSCOCO {SUB}, ASR at R@1 (%), TR / IR  (rows done so far) ══")
print("  " + f"{'source':<10}{'method':<20}" + "".join(f"{t:>16}" for t in ALL_SRC) + f"{'mean transfer':>15}")
paper_rows = []
for src in ALL_SRC:
    Rs = load_saved(src)
    tgt = [t for t in ALL_SRC if t != src]
    for lab in ["SA-AET", f"proj_L7-10_lam{LAM:g}"]:
        if lab == "SA-AET" and ("SA-AET", 42) not in Rs and src in PAPER and SUB == "full" \
                and (f"proj_L7-10_lam{LAM:g}", 42) in Rs:
            r = PAPER[src]                                       # no own baseline run: show the paper's row
            mt = sum(r[t][k] for t in tgt for k in (0, 1)) / (2 * len(tgt))
            print("  " + f"{src:<10}{'SA-AET (paper)':<20}" + "".join(f"{r[t][0]:>7.2f} /{r[t][1]:>6.2f}" for t in ALL_SRC)
                  + f"{mt:>15.2f}")
            paper_rows.append({"source": src, "method": "SA-AET (paper)", "asr": {t: r[t] for t in ALL_SRC},
                               "mean_transfer": mt})
            continue
        if (lab, 42) not in Rs or not all(t in Rs[(lab, 42)] for t in ALL_SRC):
            continue
        r = Rs[(lab, 42)]
        mt = sum(r[t][k] for t in tgt for k in (0, 1)) / (2 * len(tgt))
        name = "SA-AET" if lab == "SA-AET" else "SA-AET + proj_univ"
        print("  " + f"{src:<10}{name:<20}" + "".join(f"{r[t][0]:>7.2f} /{r[t][1]:>6.2f}" for t in ALL_SRC) + f"{mt:>15.2f}")
        paper_rows.append({"source": src, "method": name, "asr": {t: r[t] for t in ALL_SRC}, "mean_transfer": mt})
    pf = FOUT / f"paired_{src}_{SUB}_proj_L7-10_lam{LAM:g}.json"
    if pf.exists():
        d = json.load(open(pf))
        print(f"  {'':<10}{'  gain (paired)':<20}mean {d['mean_gain']:+.2f}  95% CI [{d['mean_ci'][0]:+.2f}, "
              f"{d['mean_ci'][1]:+.2f}]  p = {d['mean_p']:.4f}")
json.dump(paper_rows, open(FOUT / f"paper_table_{SUB}.json", "w"), indent=2)
print(f"  done so far: {sorted({r['source'] for r in paper_rows})} of {ALL_SRC}  -> {FOUT / f'paper_table_{SUB}.json'}")
print(f"  saved -> {FOUT}")
```
